## Instead of analysing all 158 columns individually, we create meaningful construct-level scores.

```python
materialism_cols = [
    "M1","M2","M3","M4","M5","M6","M7","M8","M9","M10",
    "M11","M12","M12B","M14","M15","M16","M16B",
    "M18","M19","M20","M21","M22","M22B"
]

df["Materialism_Score"] = df[materialism_cols].mean(axis=1)
```

```python
attitude_cols = [
    "A1","A2","A3","A4","A5","A6","A7","A8","A9"
]

df["SocialMedia_Attitude"] = df[attitude_cols].mean(axis=1)
```

```python
compulsive_cols = [
    "B1","B2","B3","B4","B5",
    "B6","B7","B8","B9","B10"
]

df["Compulsive_Buying"] = df[compulsive_cols].mean(axis=1)
```

```python
conspicuous_cols = [
    "U1","U2","U3","U4","U5",
    "U6","U7","U8","U9","U10"
]

df["Conspicuous_Buying"] = df[conspicuous_cols].mean(axis=1)
```

```python
impulse_cols = [
    "I1","I2","I3","I4","I5","I6","I7","I8","I9","I10",
    "I11","I12","I13","I14","I15","I16","I17","I18","I19",
    "I20","I21","I22","I23","I24","I25","I26","I27","I28",
    "I29","I30","I31","I32","I33","I34","I35","I36","I37","I38"
]

df["Impulse_Buying"] = df[impulse_cols].mean(axis=1)
```

```python
smi_cols = [
    "SMI1","SMI2","SMI3","SMI4",
    "SMI5","SMI6","SMI7"
]

df["SMI_Score"] = df[smi_cols].mean(axis=1)
```

# STATISTICAL ANALYSIS

### Calculate Pearson correlations

```python
variables = [
    "Materialism",
    "Social_Media_Intensity",
    "Social_Media_Content_Attitude",
    "Compulsive_Buying",
    "Conspicuous_Buying",
    "Impulse_Buying"
]

correlation_results = []

for x in variables:
    for y in variables:
        if x != y:
            r, p = pearsonr(
                analysis_df[x],
                analysis_df[y]
            )

            correlation_results.append({
                "Variable_1": x,
                "Variable_2": y,
                "Correlation": r,
                "P_Value": p
            })

correlation_df = pd.DataFrame(correlation_results)

correlation_df.head(20)
```

### Get only statistically significant relationships

```python
significant_correlations = correlation_df[
    correlation_df["P_Value"] < 0.05
].sort_values(
    by="Correlation",
    ascending=False
)

significant_correlations.head(20)
```

# KEY FINDINGS

### Identify strongest relationships

```python
corr_matrix = analysis_df.corr()

corr_pairs = (
    corr_matrix
    .where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    .stack()
    .reset_index()
)

corr_pairs.columns = ["Variable_1", "Variable_2", "Correlation"]

corr_pairs["Absolute_Correlation"] = (
    corr_pairs["Correlation"].abs()
)

corr_pairs.sort_values(
    "Absolute_Correlation",
    ascending=False
).head(10)
```

# STATISTICAL SIGNIFICANCE TABLE

```python
results = []

for i in range(len(story_vars)):
    for j in range(i + 1, len(story_vars)):

        x = story_vars[i]
        y = story_vars[j]

        r, p = pearsonr(
            analysis_df[x],
            analysis_df[y]
        )

        results.append({
            "Variable 1": x,
            "Variable 2": y,
            "Correlation (r)": round(r, 3),
            "p-value": round(p, 4),
            "Significant": "Yes" if p < 0.05 else "No"
        })

stats_table = pd.DataFrame(results)

stats_table.sort_values(
    "Correlation (r)",
    ascending=False
)
```
